# Download a public GitHub repository into the local folder

Set the repository and run **one** of the two cells. Option A needs `git`; option B needs only Python (it downloads the zip archive of the branch, without the git history).

```python
REPO   = "https://github.com/christophys/inlineHolo"   # public repository
BRANCH = ""                                       # empty: default branch
TARGET = REPO.rstrip("/").split("/")[-1]           # local folder name
```

```python
# Option A: git clone (latest commit only)
!git clone --depth 1 {"--branch " + BRANCH if BRANCH else ""} {REPO} {TARGET}
```

```python
# Option B: zip archive, Python only
import io, os, shutil, urllib.request, zipfile
owner, name = REPO.rstrip("/").split("/")[-2:]
url = f'https://github.com/{owner}/{name}/archive/{BRANCH or "HEAD"}.zip'   # HEAD = default branch
if os.path.exists(TARGET):
    print(f'{TARGET} already exists - remove it or change TARGET')
else:
    with urllib.request.urlopen(url) as r:
        z = zipfile.ZipFile(io.BytesIO(r.read()))
    top = z.namelist()[0].split('/')[0]          # REPOSITORY-<branch or sha>
    z.extractall('.')
    shutil.move(top, TARGET)
    print('downloaded to', os.path.abspath(TARGET))
```
